# Quickstart: one clip

Smoke-test the pipeline on a short clip before committing an hour of GPU time to a full match.


```python
from footy.config import Config
from footy.io.video import probe

meta = probe('../data/raw/clip.mp4')
meta
```

```python
cfg = Config.load('../configs/pipeline.yaml', '../configs/match/example_match.yaml')
cfg.match_id, len(cfg.squad_map())
```

```python
from footy.pipeline import run_pipeline

cfg.raw['io']['max_frames'] = 300
result = run_pipeline('../data/raw/clip.mp4', cfg)
result['outputs']
```

## Sanity checks

Distances per 90 should land between 9 and 12 km for outfield players. Anything far outside that means the homography or the pitch dimensions are wrong.


```python
import pandas as pd
from footy.analytics.physical import add_kinematics, summarise_player

tracks = pd.read_parquet(result['outputs']['tracks'])
summarise_player(add_kinematics(tracks))
```
